## Holistic naive

```python

SYSTEM_PROMPT = f"""
You are an expert evaluator. Your task is to rate the overall Quality of the Answer provided for the Question on a scale of 1 to 4, where 1 is the lowest Quality and 4 is the highest Quality.
"""

USER_PROMPT_TEMPLATE = """Question:
{question}

Answer: 
{answer}
"""

template_chat = {
    "system": SYSTEM_PROMPT.strip(),
    "user": USER_PROMPT_TEMPLATE.strip()}

naive_dict = create_chat_prompt_dict(name=f'naive_quality',
                                     description=f'Naive prompt that measures the Quality dimension',
                                     template_chat=template_chat,
                                     token_constraints=[
                                         "1", "2", "3", "4"],
                                     template_tags=['naive', 'quality',
                                                    'baseline', 'scale_4'],
                                     dimension_name='quality')

metadata = {'description': 'naive holistic'}

suite = PromptSuite.from_dict(naive_dict, metadata=metadata)
suite.save(suite_folder, filename=f"quality_baseline")
```

## Formative naive

```python
dimensions = ['Relevance', 'Completeness', 'Directness']

prompt_templates = []

for criterion in dimensions:
    SYSTEM_PROMPT = f"""
    You are an expert evaluator. Your task is to rate the overall {criterion} of the Answer provided for the Question on a scale of 1 to 4, where 1 is the lowest {criterion} and 4 is the highest {criterion}.
    """

    USER_PROMPT_TEMPLATE = """Question:
    {question}

    Answer: 
    {answer}
    """

    template_chat = {
        "system": SYSTEM_PROMPT.strip(),
        "user": USER_PROMPT_TEMPLATE.strip()}

    templ_dict = create_chat_prompt_dict(name=f'naive_{criterion}',
                                         description=f'Naive prompt that measures the {criterion} dimension',
                                         template_chat=template_chat,
                                         token_constraints=[
                                             "1", "2", "3", "4"],
                                         template_tags=['naive', criterion,
                                                        'baseline', 'scale_4'],
                                         dimension_name=criterion)

    prompt_templates.append(PromptTemplate.from_dict(templ_dict))

metadata = {'description': 'naive formative'}

suite = PromptSuite.from_list(prompt_templates, metadata=metadata)
suite.save(suite_folder, filename=f"formative_naive_baseline")
```

## Holistic BARS

```python
SYSTEM_PROMPT = """
You will be given a Deal Text that describes a collaboration opportunity for creators or influencers.
Your task is to provide a 'total rating' scoring how appealing, clear, and persuasive the Deal Text is for its intended audience.
Give your answer as an integer on a scale of 1 to 5, where:
    1: means the Deal Text is poorly written, unclear, or unappealing.
    2: means the Deal Text is somewhat understandable but not persuasive or engaging.
    3: means the Deal Text is clear and moderately appealing.
    4: means the Deal Text is strong, clear, and persuasive.
    5: means the Deal Text is excellent, very clear, engaging, and convincing.
"""

USER_PROMPT_TEMPLATE = """Deal Text:
{deal_text}
"""

template_chat = {
    "system": SYSTEM_PROMPT.strip(),
    "user": USER_PROMPT_TEMPLATE.strip()}

naive_dict = create_chat_prompt_dict(name=f'BARS_quality',
                                     description=f'BARS prompt that measures the Quality dimension',
                                     template_chat=template_chat,
                                     token_constraints=[
                                         "1", "2", "3", "4", "5"],
                                     template_tags=['naive', 'quality',
                                                    'baseline', 'scale_5'],
                                     dimension_name='quality')


metadata = {'description': 'BARS_holistic'}

suite = PromptSuite.from_dict(naive_dict, metadata=metadata)
suite.save(suite_folder, filename=f"quality_BARS")
```

## Holistic BARS XML

```python
SYSTEM_PROMPT = """
<instruction>
You will be given a <deal_text> which is the marketing pitch for a creator collaboration opportunity.
Note: This text is strictly the "hook" or sales pitch; specific deliverables are handled in a separate field.

Your task is to provide a 'deal_attractiveness' score, evaluating how effectively the copy convinces a creator that this product/service is worth their time and effort.
</instruction>

<rubric>
    <definition>
        *Deal Attractiveness: To what extent does the copy highlight the value of the product/service, generate excitement, and feel tailored to a creator audience (rather than a general consumer)?*
    </definition>

    <scale>
        1 = **Spammy or Low Effort**: The text is extremely vague (e.g., "DM for collab"), looks like a bot wrote it, or feels predatory/untrustworthy. It fails to explain what the product is.

        2 = **Dry / Corporate**: The text explains the product but reads like a technical manual or a generic press release. It lacks emotion or a "hook." It focuses only on features, not benefits for the creator.

        3 = **Functional**: The text is decent. It clearly states what the product is and why it's useful. It is professional but transactional—it feels like a standard job posting rather than an exciting opportunity.

        4 = **Compelling**: The copy uses engaging language that speaks the language of creators. It effectively highlights the "perk" aspect of the barter, making the product sound desirable and high-quality.

        5 = **Irresistible**: The copy creates a sense of exclusivity or high value (FOMO). It frames the barter not just as a transaction, but as a lifestyle upgrade or a unique content opportunity. The narrative is highly persuasive.
    </scale>
</rubric>
"""

USER_PROMPT_TEMPLATE = """
<deal_text>
{deal_text}
</deal_text>
"""

template_chat = {
    "system": SYSTEM_PROMPT.strip(),
    "user": USER_PROMPT_TEMPLATE.strip()}

naive_dict = create_chat_prompt_dict(name=f'BARS_XML_quality',
                                     description=f'Holistic BARS XML prompt that measures the Quality dimension',
                                     template_chat=template_chat,
                                     token_constraints=[
                                         "1", "2", "3", "4", "5"],
                                     template_tags=['BARS', 'XML', 'quality',
                                                    'baseline', 'scale_5'],
                                     dimension_name='quality')


metadata = {'description': 'BARS_XML_holistic'}

suite = PromptSuite.from_dict(naive_dict, metadata=metadata)
suite.save(suite_folder, filename=f"quality_BARS_XML")
```
