# **McGill Feedback QA**

```python
suite_folder = Path("../prompts/PromptSuites/sandbox/MCGILL_QA_FEEDBACK")
pm = PromptManager(suite_folder)
```

# Holistic naive

```python

SYSTEM_PROMPT = f"""
You are an expert evaluator. Your task is to rate the overall Quality of the Answer provided for the Question on a scale of 1 to 4, where 1 is the lowest Quality and 4 is the highest Quality.
"""

USER_PROMPT_TEMPLATE = """Question:
{question}

Answer: 
{answer}
"""

template_chat = {
    "system": SYSTEM_PROMPT.strip(),
    "user": USER_PROMPT_TEMPLATE.strip()}

naive_dict = create_chat_prompt_dict(name=f'naive_quality',
                                     description=f'Naive prompt that measures the Quality dimension',
                                     template_chat=template_chat,
                                     token_constraints=[
                                         "1", "2", "3", "4"],
                                     tags=['naive', 'quality',
                                           'baseline', 'scale_4'],
                                     dimension_name='quality')

suite = PromptSuite.from_dict(naive_dict, description='naive holistic')
suite.save(suite_folder, filename=f"quality_baseline")
```

# Formative naive

```python
dimensions = ['Relevance', 'Completeness', 'Directness']

prompt_templates = []

for criterion in dimensions:
    SYSTEM_PROMPT = f"""
    You are an expert evaluator. Your task is to rate the overall {criterion} of the Answer provided for the Question on a scale of 1 to 4, where 1 is the lowest {criterion} and 4 is the highest {criterion}.
    """

    USER_PROMPT_TEMPLATE = """Question:
    {question}

    Answer: 
    {answer}
    """

    template_chat = {
        "system": SYSTEM_PROMPT.strip(),
        "user": USER_PROMPT_TEMPLATE.strip()}

    templ_dict = create_chat_prompt_dict(name=f'naive_{criterion}',
                                         description=f'Naive prompt that measures the {criterion} dimension',
                                         template_chat=template_chat,
                                         token_constraints=[
                                             "1", "2", "3", "4"],
                                         tags=['naive', criterion,
                                               'baseline', 'scale_4'],
                                         dimension_name=criterion)

    prompt_templates.append(PromptTemplate.from_dict(templ_dict))


suite = PromptSuite.from_list(prompt_templates, description='naive formative')
suite.save(suite_folder, filename=f"formative_naive_baseline")
```
